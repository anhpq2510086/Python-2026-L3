```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df = pd.read_csv("scores (1).csv")

# Display data
print(df)

courses = ["math", "python", "database"]

# 1. Scatter plot
plt.figure(figsize=(10, 6))

for course in courses:
    plt.scatter(df["student_id"], df[course], label=course)

plt.xlabel("Student ID")
plt.ylabel("Score")
plt.title("Student Scores")
plt.legend()
plt.grid(True)
plt.savefig("scattered.png")
plt.show()

# 2. Histograms with 3 subplots
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for i, course in enumerate(courses):
    axes[i].hist(df[course].dropna(), bins=10, edgecolor="black")
    axes[i].set_title(course.capitalize() + " Scores")
    axes[i].set_xlabel("Score")
    axes[i].set_ylabel("Frequency")

plt.tight_layout()
plt.savefig("histo.png")
plt.show()

# 3. One box plot for all courses
plt.figure(figsize=(8, 6))

sns.boxplot(data=df[courses])

plt.title("Score Distribution by Course")
plt.ylabel("Score")
plt.savefig("box.png")
plt.show()

print("All plots saved successfully!")
```
