```python
USTH Advanced Programming with Python 2026
==================================

Phung Quang Anh
2510086
import students.csv as pd
from pathlib import Path


students = pd.read_csv("students.csv")

print("=" * 60)
print("PART 1: DATA MANIPULATION")
print("=" * 60)


print("\n1. First five rows:")
print(students.head())


rows, columns = students.shape
print(f"\n2. Number of rows: {rows}")
print(f"   Number of columns: {columns}")


print("\n3. Student names and GPA:")
print(students[["name", "GPA"]])


print("\n4. Students with GPA >= 3.5:")
print(students.loc[students["GPA"] >= 3.5, ["name", "major", "GPA"]]
      .sort_values("GPA", ascending=False))


print("\n5. Students sorted by GPA (highest to lowest):")
print(students.sort_values("GPA", ascending=False, na_position="last"))


print("\n6. Average GPA by major:")
print(students.groupby("major")["GPA"].mean().round(2))



print("\n" + "=" * 60)
print("PART 2: FROM RAW DATA TO USEFUL INFO")
print("=" * 60)

scores_file = Path("scores.csv")

if not scores_file.exists():
    print(
        "\nPart 2 was not run because scores.csv was not found.\n"
        "Put scores.csv in the same folder as this script, then run again."
    )
else:
    scores = pd.read_csv(scores_file)

    print("\n1. Students data preview:")
    print(students.head())
    print("\n   Scores data preview:")
    print(scores.head())


    print("\n2. Missing values in students.csv:")
    print(students.isna().sum())
    print("\n   Missing values in scores.csv:")
    print(scores.isna().sum())


    students_clean = students.dropna(subset=["student_id", "name"]).copy()
    scores_clean = scores.dropna(subset=["student_id"]).copy()

    for data in (students_clean, scores_clean):
        for col in data.select_dtypes(include="number").columns:
            data[col] = data[col].fillna(data[col].mean())
        for col in data.select_dtypes(include=["object", "string"]).columns:
            if col != "student_id" and data[col].isna().any():
                modes = data[col].mode(dropna=True)
                if not modes.empty:
                    data[col] = data[col].fillna(modes.iloc[0])

    print("\n3. Missing values after cleaning (students):")
    print(students_clean.isna().sum())
    print("\n   Missing values after cleaning (scores):")
    print(scores_clean.isna().sum())


    merged = pd.merge(
        students_clean,
        scores_clean,
        on="student_id",
        how="inner",
        suffixes=("_student", "_score")
    )
    print("\n4. Merged dataset:")
    print(merged.head())


    score_columns = [
        col for col in scores_clean.select_dtypes(include="number").columns
        if col != "student_id"
    ]

    if not score_columns:
        print(
            "\nNo numeric score columns were found in scores.csv. "
            "Check the column names and data types."
        )
    else:

        scores_clean["average_score"] = scores_clean[score_columns].mean(axis=1)
        students_scores = pd.merge(
            students_clean,
            scores_clean[["student_id", "average_score"]],
            on="student_id",
            how="inner"
        )

        print("\n5. Each student's average score:")
        print(students_scores[["student_id", "name", "major", "average_score"]]
              .sort_values("average_score", ascending=False))

        print("\n6. Top 5 students by average score:")
        print(students_scores.nlargest(5, "average_score")[
            ["student_id", "name", "major", "average_score"]
        ])

        print("\n7. Average score by major:")
        print(students_scores.groupby("major")["average_score"].mean().round(2))

        # Save useful outputs as CSV files
        merged.to_csv("merged_students_scores.csv", index=False)
        students_scores.to_csv("student_average_scores.csv", index=False)
        print("\nSaved: merged_students_scores.csv")
        print("Saved: student_average_scores.csv")
```
